# Demo 1: entanglement in four lines
Slide 8 (Demo 1 of 3), right after slide 7. Type the circuits live, everything else is ready.

```python
import pennylane as qml
from demo_utils import plot_counts, load_ibm

dev = qml.device("default.qubit")
```

### One qubit: a fair coin
*Ask the room: what will 1,000 shots look like?*

```python
@qml.set_shots(1000)
@qml.qnode(dev)
def coin():
    qml.Hadamard(wires=0)
    return qml.counts()

plot_counts(coin(), title="One qubit after H: 1,000 shots")
```

### Two qubits, no entanglement: two independent coins

```python
@qml.set_shots(1000)
@qml.qnode(dev)
def two_coins():
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)
    return qml.counts()

plot_counts(two_coins(), title="Two independent coins")
```

### Add one CNOT: the Bell state
*Ask the room again: which bars disappear?*

```python
@qml.set_shots(1000)
@qml.qnode(dev)
def bell():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.counts()

print(qml.draw(bell)())
plot_counts(bell(), title="Bell state: always 00 or 11")
```

> Each qubit alone is still a fair coin. Together they always agree: that correlation is entanglement.
