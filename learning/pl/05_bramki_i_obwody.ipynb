# Lekcja 5. Bramki i obwody

**Cel:** zrozumieć, jak zmienia się stan między przygotowaniem a pomiarem: dlaczego bramki to macierze unitarne, co robią na sferze Blocha i jak to się ma do fizyki.

## 1. Bramka to macierz unitarna

Bramka bierze stan $v$ i zwraca $Uv$. Mechanika kwantowa stawia dwa warunki:

1. **Liniowość:** $U(\alpha e_0 + \beta e_1) = \alpha Ue_0 + \beta Ue_1$ (zasada superpozycji). Kolumny macierzy to obrazy wektorów bazy.
2. **Zachowanie normy:** $\|Uv\| = 1$ dla każdego stanu, inaczej prawdopodobieństwa nie sumowałyby się do jedynki.

Odwzorowanie liniowe zachowujące normę każdego wektora to **macierz unitarna**: $U^{*}U = I$. Wynika z tego, że:

- każda bramka jest **odwracalna**: $U^{-1} = U^{*}$ (w przeciwieństwie do klasycznego AND);
- bramka zachowuje wszystkie iloczyny skalarne, $\langle Uu, Uv\rangle = \langle u, v\rangle$, więc stany ortogonalne zostają ortogonalne, a baza ortonormalna przechodzi w bazę ortonormalną.

**Unitarność nie zależy od bazy:** zmiana bazy ortonormalnej to $U \mapsto V^{*}UV$ z unitarnym $V$, a iloczyn macierzy unitarnych jest unitarny. Bramka zachowuje normę w każdej bazie ortonormalnej, bo norma to cecha wektora, a nie współrzędnych.

Pomiar **nie** jest bramką: nie jest unitarny ani odwracalny. Zwykle to osobny, ostatni krok (bywa też pomiar w środku obwodu, np. przy aktywnym resecie i korekcji błędów).

## 2. Trzy podstawowe bramki

$$X = \begin{pmatrix}0 & 1\\ 1 & 0\end{pmatrix}, \qquad Z = \begin{pmatrix}1 & 0\\ 0 & -1\end{pmatrix}, \qquad H = \frac{1}{\sqrt2}\begin{pmatrix}1 & 1\\ 1 & -1\end{pmatrix}.$$

- **X** (kwantowy NOT): $X(\alpha, \beta) = (\beta, \alpha)$, zamienia amplitudy.
- **Z** (odwrócenie fazy): $Z(\alpha, \beta) = (\alpha, -\beta)$. Nie zmienia modułów, więc $P(0)$ i $P(1)$ zostają; zmienia fazę względną o $\pi$: $Zu_+ = u_-$.
- **H** (Hadamard): $He_0 = u_+$, $He_1 = u_-$, zamienia bazę $z$ z bazą $x$. To znormalizowana macierz Hadamarda: bez czynnika $1/\sqrt2$ nie byłaby unitarna.

## 3. Na sferze Blocha bramka to obrót

Bramka zachowuje iloczyny skalarne, więc na sferze nie ściska i nie rozciąga: jest **obrotem**.

| Bramka | Obrót | Efekt |
|---|---|---|
| X | o $\pi$ wokół osi $x$ | bieguny się zamieniają, $u_\pm$ zostają |
| Z | o $\pi$ wokół osi $z$ | $u_+ \leftrightarrow u_-$ po równiku, bieguny zostają |
| H | o $\pi$ wokół osi $(x + z)/\sqrt2$ | $e_0 \leftrightarrow u_+$, $e_1 \leftrightarrow u_-$ |

Formalnie to homomorfizm $SU(2) \to SO(3)$, dwukrotne nakrycie (stąd znowu połówki kątów).

**Pomiar w innej bazie:** sprzęt mierzy tylko w $\{e_0, e_1\}$. Żeby zmierzyć w $\{u_+, u_-\}$, nakłada się $H$ przed pomiarem: $Hu_+ = e_0$, $Hu_- = e_1$, więc wynik 0 znaczy „+”, a 1 znaczy „−”.

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)
GREEN, DARK, LIGHT, GREY, ORANGE = "#44B749", "#1A8A1F", "#96E199", "#8A8A8A", "#E08A00"

e0 = np.array([1, 0], dtype=complex)   # etykieta |0⟩
e1 = np.array([0, 1], dtype=complex)   # etykieta |1⟩

def norm(v):
    return np.sqrt(np.sum(np.abs(v) ** 2))

def show(v, n=None):
    """Wypisz wektor stanu z etykietami bitów (etykieta = pozycja w zapisie dwójkowym)."""
    v = np.asarray(v)
    n = n or int(np.log2(len(v)))
    for k, a in enumerate(v):
        if abs(a) > 1e-12:
            print(f"  pozycja {k} ({format(k, f'0{n}b')}): amplituda {np.round(a, 3)}, P = {abs(a)**2:.3f}")

X = np.array([[0, 1], [1, 0]], dtype=complex)
Z = np.array([[1, 0], [0, -1]], dtype=complex)
H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2)
I2 = np.eye(2)
up, um = (e0 + e1) / np.sqrt(2), (e0 - e1) / np.sqrt(2)

for nazwa, U in (("X", X), ("Z", Z), ("H", H)):
    print(f"{nazwa}: unitarna? {np.allclose(U.conj().T @ U, I2)}")

print("Z u+ =", np.round(Z @ up, 3), "= u-")
print("H Z H =\n", np.round((H @ Z @ H).real, 3) + 0.0, "\n(porównaj z macierzą X)")
```

## 4. Interferencja w działaniu: $HH = I$

$$H(He_0) = Hu_+ = \frac{1}{\sqrt2}(He_0 + He_1) = \frac12\big[(e_0 + e_1) + (e_0 - e_1)\big] = e_0.$$

Do $e_1$ prowadzą dwie „ścieżki” z amplitudami $+\tfrac12$ i $-\tfrac12$: **znoszą się**. Do $e_0$ dwie ścieżki z $+\tfrac12$ i $+\tfrac12$: **dodają się**. Z monetą to niemożliwe: prawdopodobieństwa nie mogą się odejmować, amplitudy mogą. Każdy algorytm kwantowy ustawia obliczenie tak, żeby ścieżki do złych odpowiedzi się znosiły, a do dobrej sumowały.

```python
v1 = H @ e0
v2 = H @ v1
print("po jednym H:", np.round(v1, 3), "-> P =", np.round(np.abs(v1) ** 2, 3))
print("po dwóch H: ", np.round(v2, 3), "-> P =", np.round(np.abs(v2) ** 2, 3))

# ścieżki do e1: e0 -> e0 -> e1 oraz e0 -> e1 -> e1
sciezka_a = H[0, 0] * H[1, 0]
sciezka_b = H[1, 0] * H[1, 1]
print(f"amplitudy dwóch ścieżek do e1: {sciezka_a.real:+.2f} i {sciezka_b.real:+.2f}, suma {sciezka_a + sciezka_b:.2f}")
```

## 5. Obwód to iloczyn macierzy

Obwód kwantowy (program) to:

1. stan początkowy, zwykle $e_0 \otimes \dots \otimes e_0$;
2. ciąg bramek $U_1, U_2, \dots, U_k$;
3. pomiar na końcu.

Stan końcowy: $v_\text{koniec} = U_k \cdots U_2 U_1\, v_\text{start}$. **Kolejność jest od prawej**: pierwsza bramka stoi najbliżej wektora (na schematach czyta się od lewej). Cały obwód bez pomiaru to jedna macierz unitarna $U = U_k \cdots U_1$. Algorytm to wybór takiego $U$, żeby rozkład po pomiarze był skupiony na dobrej odpowiedzi.

Symulator na laptopie dosłownie mnoży wektor przez te macierze. Chip ich nie mnoży: transpiler zamienia bramki na natywne bramki procesora, a te są wykonywane jako impulsy mikrofalowe.

## 6. Skąd unitarność fizycznie i gdzie się kończy

**Fizyka:** zamknięty układ kwantowy ewoluuje zgodnie z równaniem Schrödingera, a jego rozwiązaniem jest zawsze macierz unitarna $U(t)$. Dla stałego hamiltonianu $\mathcal{H}$ (energii) $U = e^{-i\mathcal{H}t/\hbar}$; dla impulsu o zmiennym kształcie $\mathcal{H}(t)$ zmienia się w czasie i $U$ to eksponenta uporządkowana w czasie, ale nadal unitarna. Unitarność to nie wybór modelu, tylko sposób, w jaki ewoluują układy kwantowe.

**Inżynieria:** impuls mikrofalowy zmienia hamiltonian transmonu na określony czas; kształt, długość i fazę dobiera się tak, żeby wynikowa macierz $U$ była żądaną bramką.

**Energia:** impulsy wymieniają energię z kubitem w kontrolowany sposób. $e_0 \to e_1$ dodaje kwant energii, $e_1 \to e_0$ go oddaje, obroty wokół osi $z$ nie zmieniają obsadzeń wcale. Na chipach IBM bramka $Z$ jest wirtualna: oprogramowanie tylko przesuwa fazę kolejnych impulsów.

**Odpowiedniość macierze ↔ fizyka:**

- operacjom fizycznym odpowiadają macierze unitarne **z dokładnością do fazy globalnej** (grupa $PU(2^n) = U(2^n)/U(1)$);
- każdą macierz unitarną da się przybliżyć z dowolną dokładnością bramkami z małego, uniwersalnego zestawu, a twierdzenie Solovaya–Kitaeva mówi, że kosztuje to tylko polilogarytmicznie wiele bramek względem $1/\varepsilon$ (ważne dla dyskretnych zestawów z korekcją błędów, np. Clifford+T; natywne $R_z$ na IBM jest ciągłe); ogólna macierz $2^n \times 2^n$ wymaga jednak wykładniczo wielu bramek, a ciekawe algorytmy to te nieliczne, które da się złożyć tanio;
- działanie sprzętu można odtworzyć **tomografią procesu**, ale tylko dla 1–2 kubitów, bo liczba pomiarów rośnie wykładniczo;
- prawdziwy sprzęt nie jest układem zamkniętym: szum wyprowadza poza macierze unitarne, do kanałów kwantowych na macierzach gęstości. Stąd 0,76 na ibm_fez i 0,76–0,82 na ibm_marrakesh zamiast 0,945 z symulatora.

## Podsumowanie

> Bramka to macierz unitarna: zachowuje normę i iloczyny skalarne w każdej bazie ortonormalnej, jest odwracalna, a na sferze Blocha jest obrotem. Obwód to iloczyn bramek (od prawej) zastosowany do stanu startowego, a na końcu pomiar. Fizycznie bramki to impulsy mikrofalowe, których ewolucja $e^{-i\mathcal{H}t/\hbar}$ jest dokładnie tą macierzą, o ile układ jest zamknięty.
